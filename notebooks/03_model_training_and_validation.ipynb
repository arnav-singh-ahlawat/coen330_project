## Event-Aware Split

The main realistic evaluation uses this event-aware blocked split:

- Train = failure events 1 and 2
- Validation = failure event 3
- Test = failure event 4

This keeps each documented failure event entirely in one split. The held-out test event is not used for model selection, hyperparameter selection, or threshold selection.
